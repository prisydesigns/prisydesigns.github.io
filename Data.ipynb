```python
import pandas as pd

results = {
    "model": [
        "Model A", "Model A", "Model A", "Model A",
        "Model B", "Model B", "Model B", "Model B",
        "Model C", "Model C", "Model C", "Model C"
    ],
    "question": [
        "Q1", "Q2", "Q3", "Q4",
        "Q1", "Q2", "Q3", "Q4",
        "Q1", "Q2", "Q3", "Q4"
    ],
    "correct": [
        1, 1, 0, 1,
        1, 0, 0, 1,
        1, 1, 1, 1
    ]
}

df = pd.DataFrame(results)

df
```

```python
accuracy = df.groupby("model")["correct"].mean() * 100

print(accuracy)
```

```python
import matplotlib.pyplot as plt

accuracy.sort_values().plot(
    kind="bar",
    title="Accuracy by Language Model",
    ylabel="Accuracy (%)",
    xlabel="Model",
    ylim=(0, 100)
)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd

data = {
    "model": [
        "Model A", "Model A", "Model A", "Model A",
        "Model B", "Model B", "Model B", "Model B",
        "Model C", "Model C", "Model C", "Model C"
    ],
    "question_type": [
        "Reasoning", "Reasoning", "Knowledge", "Knowledge",
        "Reasoning", "Reasoning", "Knowledge", "Knowledge",
        "Reasoning", "Reasoning", "Knowledge", "Knowledge"
    ],
    "correct": [
        1, 0, 1, 1,
        0, 0, 1, 1,
        1, 1, 1, 1
    ]
}

df = pd.DataFrame(data)
df
```

```python
comparison = df.pivot_table(
    index="model",
    columns="question_type",
    values="correct",
    aggfunc="mean"
) * 100

print(comparison)
```
